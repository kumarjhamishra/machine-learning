Types of Annova (3 types)

1. one way Anova:
1 factor with atleast 2 levels , these levels are independant

eg - Doctor wants to test a new medication to decrease headache
split participants in 3 groups [10mg, 20mg, 30mg]
rate headache [1-10]
Medication -> Factor
10 mg   20mg    30mg -> levels


2. Repeated Measures Anova; 
one factor with atleast 2 levels, levels are dependant
Running -> Factor
Day1    Day2    Day3 -> Dependant levels
8       5       4

3. Factorial Anova;
2 or more factors each of which with atleast 2 levels, levels can be independant and dependant

Running -> Factor
Day1    Day2    Day3 -> dependant levels

Gender -> Factor
Male    Female -> independant levels
