## Multiplication Rule

### Independent events 
2 events are independant if they do not affect one another. 
eg - tossing a coin {1st H 2nd T}
eg - rolling a dice {1st 1 2nd 2}

### Dependant events
2 events when they affect each other
eg - taka a king card then queen card from deck
initially total = 52 cards = 4/52
after taking king card total = 51 cards = 4/51

## Multiplication rule
### Independant events
P(H and T) = P(H) * P(T)

## Dependant events
P(K and Q) = P(K) * P(Q/K) (probability of Q given K) -> conditional probability
