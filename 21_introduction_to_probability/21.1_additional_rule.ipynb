## Probability

###
1. Probability: about determining the likelihood of an event. 
eg - toss a coin {H, T}

## Mutually exclusive events
### 2 events are mutual exclusive if they can not occur at same time 
eg - tossing a coin and getting head and tail at same time

P (H or T) = P(H) + P(T) -> {Additive rule for mutual exlusive events}

## Non Mutual exlusive events
### 2 events can occur at same time

eg - taking a card from deck 
a kings card and a king of hearts card -> they can occur simultanouesly

P (K or heart) = P(K) + P(heart) - P(K and heart)
