#### Decorators are a powerful and flexible feature in Python that allows you to modify the behaviour of a function or class method. They are commonly used to add functionality to functions or methods without modifying their actual code. 

```python
## function copy
def welcome():
    return "Welcome to the advanced python course"

welcome()
```

```python
wel = welcome
print(wel())
del welcome
print(wel())
```

```python
# closures - function inside function

def main_welcome(msg):
    def sub_welcome():
        print("sub functions msg 1")
        print(msg)
        print("sub function msg 2")
    return sub_welcome()
```

```python
main_welcome('welcome')
```

```python
# closures - function inside function

def main_welcome(func):
    def sub_welcome():
        print("sub functions msg 1")
        func('Print function executing here')
        print("sub function msg 2")
    return sub_welcome()
```

```python
main_welcome(print)
```

```python
# default length function

def main_welcome(func, lst):

    def sub_welcome():
        print('sub function msg 1')
        print(func(lst))
        print('sub function msg 2')
    return sub_welcome()
```

```python
main_welcome(len, [1, 2, 3, 4, 5])
```

```python
len([1,2,3,4,5,6])
```

```python
# Decorator

def main_welcome(func):

    def sub_welcome():
        print('sub function msg 1')
        func()
        print('sub function msg 2')
    return sub_welcome()
```

```python
def course_introduction():
    print("This is an advanced python course")

course_introduction()
```

```python
main_welcome(course_introduction)
```

```python
@main_welcome
def course_introduction():
    print("This ia an advanced python course")
```

```python
## Decorator

def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called")
        func()
        print("Something is happening after the function is called")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("Hello!")
```

```python
say_hello()
```

```python
# decorator wit harguments
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator
        
```

```python
@repeat(3)
def say_hello():
    print("hello")
```

```python
say_hello()
```
