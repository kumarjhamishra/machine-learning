## Binomial Distribution

parameters n and p, success of n independeant experiments, with success p. Single experiment is called bernoulli trial. Outcome is 0 and 1. 
For discrete random variable
Outcome is binary
Experiment performed for n trials
eg - tossing coin for 10 times
Notation :- B(n,p)
Parameters: n -> number of trials 
p : [0,1] -> success probability for each trial
q: 1-p
k: number of successes

### PMF


Mean of distribution = n*p
variance = npq
