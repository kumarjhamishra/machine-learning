## Poisson Distribution
probability of a given number of events in a fixed interval of time. 
for discrete random variable
eg - no of people visiting hospital every hour
no of people visiting bank every hour
lambda -> expected no of events occuring at every time interval

### PMF

probability of person visting at 4 hour and 5 hour = P(x=4) + P(x=5)
