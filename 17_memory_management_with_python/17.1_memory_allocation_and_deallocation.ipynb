## reference counting

#### Reference counting is the primary method Python uses to manage memory. Each object in Python maintains a count of references pointing to it. When the reference count drops to zero, the memory occupied by the object is deallocated

```python
## reference counting

import sys

a = []
# 1 reference coming from vairable 1 from function
print(sys.getrefcount(a))
```

```python
b = a
print(sys.getrefcount(a))
```

```python
del b
print(sys.getrefcount(a))
```

## Garbage Collection

#### Python includes a cyclic garbage collector to handle reference cycles. Reference cycles occur when objects reference each other, preventing their reference counts from reaching zero. 

```python
import gc

# enable garbage collector
gc.enable()
```

```python
# disable
gc.disable()
```

```python
# manually trigger 
gc.collect()
```

```python
## get garbage collection stats
gc.get_stats()
```

```python
## get unreachable objects
gc.garbage
```

## Memory Management Best Practices

####
1. Use local variables - shorted lifespan and freed sooner than global
2. Avoid circular references - can lead to memory leaks
3. Use generators - generatos produce items one at a time and only keeps one item in memory at a time, making them memory efficient.
4. Explictly Delete objects - use del statement to explictly delete them
5. Profile memory usage - use memory profiling tools like tracemalloc and memory_profiler to identiy memory leaks and optimize memory usage. 
####

```python
import gc

class MyObject:
    def __init__(self, name):
        self.name = name
        print(f"Object {self.name} created")

    def __del__(self):
        print(f"Object {self.name} deleted")

# create circular reference
obj1 = MyObject("obj1")
obj2 = MyObject("obj2")
obj1.ref = obj2
obj2.ref = obj1


# will not be called due to circula reference
del obj1
del obj2

# manually trigger the garbage collection
gc.collect()
```

```python
# generators for memory efficiency

def generate_numbers(n):
    for i in range(n):
        yield i

for num in generate_numbers(1000000):
    print(num)
    if num > 10:
        break
```

```python
# Profiling memroy usage with tracemalloc

import tracemalloc

def create_list():
    return [i for i in range(100000)]

def main():
    tracemalloc.start()

    create_list()

    snapshot = tracemalloc.take_snapshot()
    top_stats = snapshot.statistics('lineno')

    print("[ Top 10 ]")
    for stat in top_stats[:10]:
        print(stat)
```

```python
main()
```
