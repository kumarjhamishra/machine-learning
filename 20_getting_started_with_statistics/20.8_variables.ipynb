## Variable

### Variable property can take up any value. 

1. Quantitative Variable:
1.1 Discrete quantitative
eg - how many children does A have ? 
whole number , no fraction

1.2 Continuous quantitative
eg - height = 175.3 cm
can have any value 

2. Qualitative Variable:
values and not numbers
Gender - male, female
Color - red, green, blue
