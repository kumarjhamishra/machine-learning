## Random Variable

### Denoted by capital X 
it is a function whose value is derived from different process or experiments

eg - tossing a coin: head (0 value) or tail (1 value)
rolling a fair dice - {1,2,3,4,5,6}

## Different types of Random Variables
### 1. Discrete    2. Continous

Discrete - tossing a coin, rolling a dice

Continous - 
eg - tomorrow how many inches it is going to rain ? any value
0, 1.1 inch, 4.3 inch, 5.inch

eg - height of people attending event tomorrow.
