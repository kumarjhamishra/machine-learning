```python
import seaborn as sns
```

```python
# basic plottting with seaborn

tips = sns.load_dataset('tips')
tips
```

```python
# create a scatter plot
import matplotlib.pyplot as plt

sns.scatterplot(data=tips, x='total_bill', y='tip')
plt.title('Scatter Plot of Total Bill vs Tip')
plt.show()
```

```python
# line plot
sns.lineplot(data=tips, x='size', y='total_bill')
plt.title('Total Bill by size')
plt.show()
```

```python
# categorical plots

# bar plot
sns.barplot(data=tips, x='day', y='total_bill')
plt.title('Total Bill vs Day')
plt.show()
```

```python
# box plot

sns.boxplot(data=tips, x='day', y='total_bill')
```

```python
# violin plot
sns.violinplot(data=tips, x='day', y='total_bill')
```

```python
# histogram
sns.histplot(tips['total_bill'], bins=10, kde=True)
```

```python
# KDE plot
sns.kdeplot(tips['total_bill'], shade=True)
```

```python
# pairplot - relation b/w each and every variables
sns.pairplot(data=tips)
```

```python
# Heatmap

corr = tips.corr()
corr
```

```python
# Heatmap
# for numerical cols
corr = tips[['total_bill', 'tip', 'size']].corr()
corr 
```

```python
sns.heatmap(corr, annot=True, cmap='coolwarm')
```

```python
import pandas as pd

sales_df = pd.read_csv('sales_data.csv')
sales_df.head(5)
```

```python
# plot total sales by product

plt.figure(figsize=(10,6))
sns.barplot(data=sales_df, x='Product Category', y='Total Revenue', estimator=sum)
plt.title('Total Sales by Product')
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.show()
```

```python
# plot total sales by region

plt.figure(figsize=(10,6))
sns.barplot(data=sales_df, x='Region', y='Total Revenue', estimator=sum)
plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.show()
```
