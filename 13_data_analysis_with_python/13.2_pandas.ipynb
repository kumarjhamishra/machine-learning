```python
import pandas as pd
import numpy as np
```

```python
## series
## panda series is 1d array like object, similar to a column in table

data = [1,2,3,4,5,6]
series=pd.Series(data)
print("series \n", series)

# left coulmn of o/p is the default indices
print(type(series))
```

```python
# create a series from dictinoary elements
data = {'a': 1, 'b': 2, 'c': 3, 'aatmaram': 5}
series_dict = pd.Series(data)
print(series_dict) # default index has changed, ' keys will become the index
# pandas uses the order how the elements are inserted in dictionary
```

```python
# giving specific index order
data = [1,2,3]
index = ['z', 'm', 'o']
pd.Series(data, index=index)
```

```python
# dataframe - it is 2dimensional, which can contain heterogenuous data

# create a dataframe from a dictionary of list
data = {
    'Name': ['Ramesh', 'Suresh', 'Mahesh'],
    'Age': [20, 52, 26],
    'City': ['Pune', 'Gurgaon', 'Assam']
}


df = pd.DataFrame(data)
print(df)
print(type(df))
```

```python
np.array(df)
```
