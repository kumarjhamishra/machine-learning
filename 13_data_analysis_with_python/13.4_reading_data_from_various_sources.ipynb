```python
import requests
from io import StringIO

url = "https://en.wikipedia.org/wiki/Mobile_country_code"
headers = {
    "User-Agent":"Mozilla/5.0"
}
response = requests.get(url, headers=headers, timeout=30)
response.raise_for_status()

tables = pd.read_html(
    StringIO(response.text),
    match="Country",
    header=0
)

tables
```
