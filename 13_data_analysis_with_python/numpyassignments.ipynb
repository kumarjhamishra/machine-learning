```python
# 1 - Assignment 1: Array Creation and Manipulation

# 1. Create a NumPy array of shape (5, 5) filled with random integers between 1 and 20. Replace all the elements in the third column with 1.


import numpy as np

arr = np.random.randint(1, 21, size=(5,5))

print("Original array: ")
print(arr)

arr[:,2] = 1

print("Moodified array: ")
print(arr)
```

```python
# 2. Create a NumPy array of shape (4, 4) with values from 1 to 16. Replace the diagonal elements with 0.
arr = np.arange(1, 17).reshape(4,4)
print("Original arr: ")
print(arr)

# replace diagonal elements with 0
np.fill_diagonal(arr, 0)
print("Modified arr: ")
print(arr)
```

```python
# Assignment 2: Array Indexing and Slicing

# 1. Create a NumPy array of shape (6, 6) with values from 1 to 36. Extract the sub-array consisting of the 3rd to 5th rows and 2nd to 4th columns.
arr = np.arange(1, 37).reshape(6,6)
print("Original arr: \n", arr)

sub_arr = arr[2:5, 1:4]
print("Sub array: \n", sub_arr)
```

```python
# 2. Create a NumPy array of shape (5, 5) with random integers. Extract the elements on the border.
arr = np.random.randint(1,26, size=(5,5))
print("Original arr: \n", arr)

# border elements
border_arr = np.concatenate((arr[0, :], arr[-1, :], arr[1:-1, 0], arr[1:-1, -1]))
print("Border elements: \n", border_arr)
```

```python
# Assignment 3: Array Operations

# 1. Create two NumPy arrays of shape (3, 4) filled with random integers. Perform element-wise addition, subtraction, multiplication, and division.
arr1 = np.random.randint(1, 13, size=(3,4))
arr2 = np.random.randint(1, 13, size=(3,4))
print("arr1: \n", arr1)
print("arr2: \n", arr2)

print("addition: \n", arr1+arr2)
print("subtraction: \n", arr1-arr2)
print("multiplication: \n", arr1*arr2)
print("division: \n", arr1/arr2)
```

```python
# 2. Create a NumPy array of shape (4, 4) with values from 1 to 16. Compute the row-wise and column-wise sum.
arr = np.arange(1, 17).reshape(4,4)

print("Arr: \n", arr)

print("Row wise sum")
print(np.sum(arr, axis=1))
print("Col wise sum")
print(np.sum(arr, axis=0))
```

```python
# Assignment 4: Statistical Operations

# 1. Create a NumPy array of shape (5, 5) filled with random integers. Compute the mean, median, standard deviation, and variance of the array.
arr = np.random.randint(1, 26, size=(5,5))
print("Mean: ", np.mean(arr))
print("Median: ", np.median(arr))
print("Standard Deviation: ", np.std(arr))
print("Variance: ", np.var(arr))
```

```python
# 2. Create a NumPy array of shape (3, 3) with values from 1 to 9. Normalize the array (i.e., scale the values to have a mean of 0 and a standard deviation of 1).
arr = np.arange(1, 10).reshape(3,3)
print("Arr: \n", arr)

# normalize the array
mean = np.mean(arr)
std_dev = np.std(arr)
normalized = (arr - mean)/std_dev

print("Normalized array: \n", normalized)
print("Mean of normalized: ", np.mean(normalized))
print("Standard deviation: ", np.std(normalized))
```

```python
# Assignment 5: Broadcasting

# 1. Create a NumPy array of shape (3, 3) filled with random integers. Add a 1D array of shape (3,) to each row of the 2D array using broadcasting.
arr = np.random.randint(1, 10, size=(3,3))
row_arr = np.random.randint(1, 10, size=(3,))
print("arr: \n", arr)
print("row_arr: \n", row_arr)
new_arr = arr+row_arr
print("new: \n", new_arr)
```

```python
# 2. Create a NumPy array of shape (4, 4) filled with random integers. Subtract a 1D array of shape (4,) from each column of the 2D array using broadcasting.
arr = np.random.randint(1, 17, size=(4,4))
print("arr: \n", arr)

row_arr = np.random.randint(1, 17, size=(4,))

print("1-d arr: \n", row_arr)

new_arr = arr - row_arr[:, np.newaxis]
print("new_arr: \n", new_arr)
```

```python
# Assignment 6: Linear Algebra

# 1. Create a NumPy array of shape (3, 3) representing a matrix. Compute its determinant, inverse, and eigenvalues.
arr = np.random.randint(1, 10, size=(3,3))
print("arr: \n", arr)

determinant = np.linalg.det(arr)
inverse = np.linalg.inv(arr)
eigen_values = np.linalg.eigvals(arr)
print("Determinant: ", determinant)
print("Inverse: ", inverse)
print("Eigen value: ", eigen_values)
```

```python
# 2. Create two NumPy arrays of shape (2, 3) and (3, 2). Perform matrix multiplication on these arrays.
arr1 = np.random.randint(1, 5, size=(2, 3))
print("arr1: \n", arr1)

arr2 = np.random.randint(1, 5, size=(3, 2))
print("arr2: \n", arr2)

arr3 = np.dot(arr1, arr2)
print("arr3: \n", arr3)
```

```python
# Assignment 7: Advanced Array Manipulation

# 1. Create a NumPy array of shape (3, 3) with values from 1 to 9. Reshape the array to shape (1, 9) and then to shape (9, 1).
arr = np.arange(1, 10).reshape(3,3)
print("Original arr: \n", arr)

print("Shape (1,9): \n", arr.reshape(1, 9))

print("Shape (9,1): \n", arr.reshape(9,1))
```

```python
# Assignment 9: Structured Arrays

# 1. Create a structured array with fields 'name' (string), 'age' (integer), and 'weight' (float). Add some data and sort the array by age.
arr_dtype = np.dtype([
    ('Name', 'U20'), # unicode string of max 20 char
    ('Age', 'i4'), # integer of 4 byte
    ('Weight', 'f4') # float of 4 byte
])

arr = np.array([
    ('Ram', 20, 70.9),
    ('Shyam', 22, 65.3),
    ('Gita', 18, 60.0)
], dtype=arr_dtype)

print("Arr: \n", arr)
print("Sorting arr on age feild: \n", np.sort(arr, order='Age'))
```

```python
# 2. Create a structured array with fields 'x' and 'y' (both integers). Add some data and compute the Euclidean distance between each pair of points.
arr_dtype = np.dtype([
    ('x', 'i4'),
    ('y', 'i4')
])

arr = np.array([
    (2, 3),
    (7, 2),
    (1, 1),
    (4, 5),
], dtype=arr_dtype)

print("Arr: \n", arr)


# But you need every point compared with every other point.

# Conceptually, you need something like:

#         P0     P1     P2     P3
# P0      0      d01    d02    d03
# P1      d10    0      d12    d13
# P2      d20    d21    0      d23
# P3      d30    d31    d32    0
# sqrt((x1-x2)**2 + (y1-y2)**2)
eucid_distance = np.sqrt((arr['x'] - arr['x'][:, np.newaxis]) ** 2 + (arr['y'] - arr['y'][:, np.newaxis])**2)
print("Euclidean distance between points: \n", eucid_distance)
```

```python
# 2. Create a masked array of shape (3, 3) with random integers and mask the diagonal elements. Replace the masked elements with the mean of the unmasked elements.
arr = np.random.randint(1, 10, size=(3,3))
print("Arr: \n", arr)

# create mask - identity matrix 3x3
mask = np.eye(3, dtype=bool)
print(mask)

# masked array
masked_arr = np.ma.masked_array(arr, mask=mask)
print("Masked arr: \n", masked_arr)

mean = masked_arr.mean() # it will only consider non masked element both in num and denum
print("Mean: ", mean)

# filled() means:

# Create a normal array where the masked positions are replaced with the value I provide.
final = masked_arr.filled(mean)
print("Final arr: \n", final) # arr is int type theirfore in o/p it dropped decimal part
```
