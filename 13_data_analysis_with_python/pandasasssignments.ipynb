```python
# Assignment 1: DataFrame Creation and Indexing

# 1. Create a Pandas DataFrame with 4 columns and 6 rows filled with random integers. Set the index to be the first column.
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 25, size=(6,4)), columns=['A','B','C','D'])
print("Original: \n", df)

# set first col to be the index for the rows
modified_df = df.set_index('A')
print("Modified arr: \n", modified_df)
```

```python
# 2. Create a Pandas DataFrame with columns 'A', 'B', 'C' and index 'X', 'Y', 'Z'. Fill the DataFrame with random integers and access the element at row 'Y' and column 'B'.
# size tells numpy how many random numbers to generate
df = pd.DataFrame(np.random.randint(1, 10, size=(3,3)), index=['X','Y','Z'], columns=['A','B','C'])
print(df)

# accessing element (row = Y, col=B)
print("Element ar row Y, col B: ", df.at['Y','B'])
```

```python
# Assignment 2: DataFrame Operations

# 1. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Add a new column that is the product of the first two columns.
df = pd.DataFrame(np.random.randint(1, 16, size=(5,3)))
print("Original: \n", df)

new_col = df[0] * df[1]
print("New columns: \n", new_col)

df['New Col'] = new_col
print("Upated df: \n", df)
```

```python
# 2. Create a Pandas DataFrame with 3 columns and 4 rows filled with random integers. Compute the row-wise and column-wise sum.
df = pd.DataFrame(np.random.randint(1, 13, size=(4,3)))
print("Original: \n", df)

row_wise_sum = df.sum(axis=1)
print("Row wise sum: \n", row_wise_sum)

col_wise_sum = df.sum(axis=0)
print("Col wise sum: \n", col_wise_sum)
```

```python
# Assignment 3: Data Cleaning

# 1. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Introduce some NaN values. Fill the NaN values with the mean of the respective columns.
df = pd.DataFrame(np.random.randint(1, 16, size=(5,3)))
print("Original: \n", df)

# some nan values
df.iloc[0, 1] = np.nan
df.iloc[2, 2] = np.nan
df.iloc[4, 0] = np.nan
print("Filled with some nan values: \n", df)

mean = df.mean(skipna=True)
print("Mean of df column's: \n", mean)

df.fillna(mean, inplace=True)
```

```python
# 2. Create a Pandas DataFrame with 4 columns and 6 rows filled with random integers. Introduce some NaN values. Drop the rows with any NaN values.
df = pd.DataFrame(np.random.randint(1, 25, size=(6,4)))
print("Original: \n", df)

# some nan values
df.iloc[0, 1] = np.nan
df.iloc[2, 2] = np.nan
df.iloc[4, 0] = np.nan
print("Filled with some nan values: \n", df)

df.dropna(axis=0, inplace=True)
print("Dropped rows with nan values: \n", df)
```

```python
# Assignment 4: Data Aggregation

# 1. Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category.
df = pd.DataFrame({'Category': np.random.choice(['A','B','C'], size=10), 'Value': np.random.randint(20, 100, size=10)})
print("Df: \n", df)

# sum and mean of each category
# only consider value column
sum_mean = df.groupby(by='Category')['Value'].agg(['sum', 'mean'])
print("Sum and mean of each category: \n", sum_mean)
```

```python
# 2. Create a Pandas DataFrame with 3 columns: 'Product', 'Category', and 'Sales'. Fill the DataFrame with random data. Group the DataFrame by 'Category' and compute the total sales for each category.
df = pd.DataFrame({'Product': np.random.choice(['P1','P2','P3'], size=10), 'Category': np.random.choice(['C1','C2','C3'], size=10), 'Sales': np.random.randint(20, 100, size=10)})
print("Original dataframe: \n", df)

total_sales = df.groupby(by='Category')['Sales'].sum()
print("Total sales for each category: \n", total_sales)
```

```python
# Assignment 5: Merging DataFrames

# 1. Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.
df1 = pd.DataFrame({'Key': ['A','B','C','D'], 'Value1': np.random.randint(1, 100, size=4)})
df2 = pd.DataFrame({'Key': ['A','B','M','E'], 'Value2': np.random.randint(1, 100, size=4)})

print("Original df1: \n", df1)
print("\nOriginal df2: \n", df2)

# merging dataframes
df3 = pd.merge(df1, df2,on='Key', how='outer')
df3
```

```python
# 2. Create two Pandas DataFrames with different columns. Concatenate the DataFrames along the rows and along the columns.
df1 = pd.DataFrame({'A': np.random.randint(1, 100, size=3), 'B': np.random.randint(1, 100, size=3)})
df2 = pd.DataFrame({'C': np.random.randint(1, 100, size=3), 'D': np.random.randint(1, 100, size=3)})


print("Original df1: \n", df1)
print("\nOriginal df2: \n", df2)


# pd.concat() never joins data by position alone. It aligns using labels:

# axis=0 (row-wise): aligns by column names. If one DataFrame is missing a column, Pandas fills those cells with NaN.
# axis=1 (column-wise): aligns by row indexes. If one DataFrame is missing an index, Pandas fills those cells with NaN.
df3 = pd.concat([df1, df2], axis=0)
print("\nConcat along rows: \n", df3)

df4 = pd.concat([df1, df2], axis=1)
print("\nConcat along cols: \n", df4)
```

```python
# Assignment 6: Time Series Analysis

# 1. Create a Pandas DataFrame with a datetime index and one column filled with random integers. Resampledf  the DataFrame to compute the monthly mean of the values.
dates = pd.date_range(start='2026-01-01', end='2026-12-31', freq='D')
print("Dates: \n", dates)

df = pd.DataFrame({'Date': dates, 'Data': np.random.randint(1, 100, size=(len(dates)))})
df.set_index(keys='Date', inplace=True)
print("\nDf: \n", df)

# monthly mean of values
monthly_mean = df.resample('ME').mean()
print("\n Monthly mean: \n", monthly_mean)
```

```python
# 2. Create a Pandas DataFrame with a datetime index ranging from '2021-01-01' to '2021-12-31' and one column filled with random integers. Compute the rolling mean with a window of 7 days.
dates = pd.date_range(start='2026-01-01', end='2026-12-31', freq='D')
print("Dates: \n", dates)

df = pd.DataFrame({'Date': dates, 'Data': np.random.randint(1, 100, size=(len(dates)))})
df.set_index('Date', inplace=True)
print("\nDf: \n", df)

# rolling mean with window of 7 days
rolling_mean = df.rolling(window=7).mean()
print("\nRolling mean with a window of 7 days: \n", rolling_mean)
```

```python
# Assignment 7: MultiIndex DataFrame

# 1. Create a Pandas DataFrame with a MultiIndex (hierarchical index). Perform some basic indexing and slicing operations on the MultiIndex DataFrame.
arrays = [['A','A','B','B'], ['one', 'two', 'one', 'two']]
print("Arrays: \n", arrays)

index = pd.MultiIndex.from_arrays(arrays=arrays, names=('Category', 'Subcategory'))
print("\nIndex: \n", index)

df = pd.DataFrame(np.random.randint(1, 100, size=(4,3)), index=index, columns=['Value1', 'Value2', 'Value3'])
print("\nDf: \n", df)

print("\nParent index A: \n", df.loc['A'])

print("\nOne particular row: \n", df.loc[('B', 'two')])
```

```python
# 2. Create a Pandas DataFrame with MultiIndex consisting of 'Category' and 'SubCategory'. Fill the DataFrame with random data and compute the sum of values for each 'Category' and 'SubCategory'.
arrays = [['A','A','B','B','C','C'], ['one', 'two', 'one', 'two', 'one', 'two']]
print("Arrays: \n", arrays)

index = pd.MultiIndex.from_arrays(arrays=arrays, names=('Category', 'Subcategory'))

df = pd.DataFrame(np.random.randint(1, 100, size=(6, 3)), index=index, columns=['V1', 'V2', 'V3'])
print("\n Df: \n", df)

# sum of values for each category and subcategory
df['Total'] = df.sum(axis=1)
print("\nCategory and subcategory wise sum: \n", df)
```

```python
# Assignment 8: Pivot Tables

# 1. Create a Pandas DataFrame with columns 'Date', 'Category', and 'Value'. Create a pivot table to compute the sum of 'Value' for each 'Category' by 'Date'.
date_range = pd.date_range(start='2026-08-17', end='2026-08-27', freq='D')
print("Date ranges: \n", date_range)

df = pd.DataFrame({'Date': np.random.choice(date_range, size=20), 'Category': np.random.choice(['A', 'B', 'C'], size=20), 'Value': np.random.randint(1, 100, size=20)})
print("\nOriginal dataframe: \n", df)

# making a pivot table
pivot = df.pivot_table(values='Value', index='Date', columns='Category', aggfunc='sum')
print("\nPrinting pivot table: \n", pivot)

```

```python
# 2. Create a Pandas DataFrame with columns 'Year', 'Quarter', and 'Revenue'. Create a pivot table to compute the mean 'Revenue' for each 'Quarter' by 'Year'.
df = pd.DataFrame({'Year': np.random.choice([2024, 2025, 2026], size=12), 'Quarter': np.random.choice(['Q1', 'Q2', 'Q3', 'Q4'], size=12), 'Revenue': np.random.randint(1, 100, size=12)})
print("Original dataframa: \n", df)

# pivot table
pivot = df.pivot_table(values='Revenue', index='Year', columns='Quarter', aggfunc='mean')
print("\nMean revenue for each quarter by year: \n", pivot)
```

```python
# Assignment 9: Applying Functions

# 1. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Apply a function that doubles the values of the DataFrame.
df = pd.DataFrame(np.random.randint(1, 100, size=(5,3)), columns=['A','B','C'])
print(type(df))
print("Original dataframe: \n", df)

import pandas as pd
print(pd.__version__)
print(hasattr(pd.DataFrame, "applymap"))
print(hasattr(df, "applymap"))
print(df.__class__)

# applymap applies the lambda fuction on each cell of dataframe
doubled_df = df.applymap(lambda x : x*2)
print("\nDoubed value: \n", doubled_df)
```
