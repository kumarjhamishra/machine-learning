```python
import sqlite3
```

```python
# conncet to an sqlite database
conn = sqlite3.connect('example.db')
```

```python
cursor = conn.cursor()
```

```python
# create a table
cursor.execute('''
    CREATE TABLE IF NOT EXISTS employee(
        id INTEGER PRIMARY KEY,
        name TEXT NOT NULL,
        age INTEGER,
        department TEXT
    )
''')

conn.commit()
conn.close()
```

```python
conn = sqlite3.connect('example.db')
cursor = conn.cursor()

cursor.execute('''
    SELECT * FROM employee
''')
```

```python
# insert the data into sqlite3 table
cursor.execute('''
    INSERT INTO employee(id, name, age, department) 
    VALUES
        (1, 'Krish', 32, 'Data Scientist'),
        (2, 'Bob', 25, 'Engineering'),
        (3, 'Charlie', 35, 'Finance')
''')

conn.commit()
```

```python
# query the data from table
cursor.execute('''
    SELECT * FROM employee
''')

records = cursor.fetchall()
for record in records:
    print(record)
```

```python
# update the data in the table
cursor.execute('''
    UPDATE employee 
    SET age = 34 
    WHERE name = 'Krish'
''')

conn.commit()
```

```python
# query the data from table
cursor.execute('''
    SELECT * FROM employee
''')

records = cursor.fetchall()
for record in records:
    print(record)
```

```python
# delete the data from table
cursor.execute('''
    DELETE FROM employee
    WHERE name = 'Bob'
''')

conn.commit()
```

```python
# query the data from table
cursor.execute('''
    SELECT * FROM employee
''')

records = cursor.fetchall()
for record in records:
    print(record)

conn.close()
```

```python
# working with sales data
conn = sqlite3.connect('sales_data.db')
cursor = conn.cursor()

cursor.execute('''
    CREATE TABLE IF NOT EXISTS sales(
        id INTEGER PRIMARY KEY,
        date TEXT NOT NULL,
        product TEXT NOT NULL,
        sales INTEGER,
        region TEXT
    )
''')

sales_data  = [
    ('2026-01-01', 'P1', 100, 'North'),
    ('2026-01-02', 'P2', 200, 'South'),
    ('2026-01-03', 'P1', 150, 'East'),
    ('2026-01-04', 'P3', 250, 'West'),
    ('2026-01-05', 'P2', 300, 'North')
]

cursor.executemany('''
    INSERT INTO sales (date, product, sales, region)
    VALUES (?,?,?,?)
''', sales_data)


conn.commit()
```

```python
# query the data from table
cursor.execute('''
    SELECT * FROM sales
''')

records = cursor.fetchall()
for record in records:
    print(record)

```
