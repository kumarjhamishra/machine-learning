```python
# Creating and Connecting to a Database

# 1. Write a Python function to create a new SQLite3 database named `test.db`.
import sqlite3 

def create_database():
    conn = sqlite3.connect('test.db')
    conn.close()
    print("Database created and successfully connected")

create_database()
```

```python
# 2. Write a Python function to create a table named `employees` with columns `id` (integer), `name` (text), `age` (integer), and `department` (text) in the `test.db` database.
def create_table():
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS employees(
            id INTEGER PRIMARY KEY,
            name TEXT NOT NULL,
            age INTEGER,
            department TEXT
        )
    ''')

    conn.commit()
    conn.close()
    print("Table employees created successfully")

create_table()
```

```python
# Assignment 3: Querying Data

# 1. Write a Python function to fetch and display all records from the `employees` table.
def fetch_from_employees():
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()

    cursor.execute('''
        SELECT * FROM employees
    ''')

    rows = cursor.fetchall()
    for row in rows:
        print(row)

    
    # conn.commit() not needed for fetch 
    conn.close()

fetch_from_employees()
```

```python
# 2. Write a Python function to fetch and display all employees from a specific department.
def fetch_from_department(department):
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()

    cursor.execute('''
        SELECT * FROM employees WHERE department = ?
    ''', (department,)) # comma after department bcoz cursor expects 2nd arg as tuple

    rows = cursor.fetchall()
    for row in rows:
        print(row)

    
    # conn.commit() not needed for fetch 
    conn.close()

fetch_from_department('HR')
```

```python
# 2. Write a Python function to fetch and display employees whose names start with a specific letter.
def fetch_employees_by_names(ch):
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()

    cursor.execute('''
        SELECT * FROM employees
        WHERE name LIKE ?
    ''', (ch + '%',))

    rows = cursor.fetchall()
    print("Employee records whose name starts with: ", ch)
    for row in rows:
        print(row)

    conn.close()

fetch_employees_by_names('A')
```

```python
# Assignment 7: Handling Transactions

# 1. Write a Python function to insert multiple employees into the `employees` table in a single transaction. Ensure that if any insertion fails, none of the insertions are committed.
def inset_multiple_employees(employees):
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()

    try:
        cursor.executemany('''
            INSERT INTO employees (id, name, age, department)
            VALUES (?,?,?,?)
        ''', employees)
        conn.commit()
        print("All employees inserted successfully")
    except Exception as e:
        conn.rollback()
        print("Error occured, transaction rolled back")
        print(e)
    finally:
        conn.close()


employees = [
    (6, 'Frank', 40, 'Finance'),
    (7, 'Grace', 29, 'Engineering'),
    (8, 'Hannah', 35, 'Marketing'),
    (9, 'Ivan', 38, 'Sales'),
    (6, 'Jack', 45, 'HR')  # Duplicate ID to cause an error
]

inset_multiple_employees(employees)
```

```python
# 2. Write a Python function to update the age of multiple employees in a single transaction. Ensure that if any update fails, none of the updates are committed.
def update_employees_age(updates):
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()

    # update does not throw error when it does not found the row
    try:
        for age, emp_id in updates:
            cursor.execute('''
                UPDATE employees
                SET age = ?
                WHERE id = ?
            ''', (age, emp_id))

            if cursor.rowcount == 0:
                raise ValueError(
                    f"Employee with ID {emp_id} does not exist"
                )
        conn.commit()
        print("Age of employees updated successfully")
    except Exception as e:
        conn.rollback()
        print("Updation of employee's age failed due to: ", e)
    finally:
        conn.close()


updates = [
    (32, 1),
    (26, 2),
    (33, 3),
    (41, 4),  # Non-existing ID to cause an error
    (23, 5)
]
update_employees_age(updates)
```

```python
# Assignment 8: Creating Relationships

# 1. Create a new table named `departments` with columns `id` (integer) and `name` (text).
conn = sqlite3.connect('test.db')
cursor = conn.cursor()

cursor.execute('''
    CREATE TABLE IF NOT EXISTS departments(
        id INTEGER PRIMARY KEY,
        name TEXT NOT NULL
    )
''')
conn.commit()
conn.close()
print("Table department created successfully")
```

```python
# 2. Modify the `employees` table to include a foreign key referencing the `id` column in the `departments` table.

conn = sqlite3.connect('test.db')
cursor = conn.cursor()

# SQLite does not support adding a foreign key constraint to an existing table with ALTER TABLE.
# It only lets you add the column.
# Foreign key enforcement is disabled by default in SQLite.
# Before using foreign keys, enable them:
cursor.execute("PRAGMA foreign_keys = ON")

# create a new table - bcoz sqlite can't add foreign key to an existing table
cursor.execute('''
    CREATE TABLE employees_new(
        id INTEGER PRIMARY KEY,
        name TEXT NOT NULL,
        age INTEGER,
        department TEXT,
        d_id INTEGER,
        FOREIGN KEY(d_id) REFERENCES departments(id)
    )
''')


# copy existing data
cursor.execute('''
    INSERT INTO employees_new(id, name, age, department)
    SELECT id, name, age, department
    FROM employees
''')

# delete old table
cursor.execute('''
    DROP TABLE employees
''')

# rename the new table
cursor.execute('''
    ALTER TABLE employees_new
    RENAME TO employees
''')

conn.commit()
conn.close()
print("Table employees add with foreign key")
```

```python
# 3. Write a Python function to insert data into both the `departments` and `employees` tables, ensuring referential integrity.
def insert_into_departments_employees(d_id, d_name, e_id, name, age, department):
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()

    try:
        cursor.execute('''
            INSERT INTO departments (id, name)
            VALUES (?,?)
        ''', (d_id, d_name))
        cursor.execute('''
            INSERT INTO employees (id, name, age, department, d_id)
            VALUES (?,?,?,?,?)
        ''', (e_id, name, age, department, d_id))
        conn.commit()
        print("Department and employee inserted successfully")
    except Exception as e:
        conn.rollback()
        print("Error encountered in inserting data: ", e)
    finally:
        conn.close()

insert_into_departments_employees(1, 'Finance', 10, 'Zara', 28, 'Finance')
```
