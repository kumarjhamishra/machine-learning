```python
# Assignment 2: Logging with Different Handlers

# 1. Write a Python function to create a logger that logs messages to both a file named `app.log` and the console.
def logging_in_file_and_console():

    # create a logger of own
    logger = logging.getLogger("my_logger")

    # set the logger's level
    logger.setLevel(logging.DEBUG)

    # create a file handler
    file_handler = logging.FileHandler('app2.log')

    # set it's level
    file_handler.setLevel(logging.DEBUG)

    # create a console handler
    console_handler = logging.StreamHandler()

    # set it's level
    console_handler.setLevel(logging.DEBUG)

    # create a formatter
    formatter = logging.Formatter(
        "%(asctime)s - %(levelname)s - %(message)s"
    )

    # give formatter to both handlers
    file_handler.setFormatter(formatter)
    console_handler.setFormatter(formatter)

    # attach handlers to the logger
    logger.addHandler(file_handler)
    logger.addHandler(console_handler)

    # log messages
    logger.debug("Debug message")

    logger.info("Info message")

    logger.warning("Warning message")

    logger.error("Error message")

    logger.critical("Critical message")


logging_in_file_and_console()
    
```

```python
# 2. Modify the function to use different logging levels for the file and console handlers.
def logging_in_file_and_console_updated():

    # create a logger of own
    logger = logging.getLogger("my_logger")

    # prevent duplicate handlers if function is run multiple times
    if logger.hasHandlers():
        logger.handlers.clear()

    # set the logger's level
    logger.setLevel(logging.DEBUG)

    # create a file handler
    file_handler = logging.FileHandler('app2.log')

    # set it's level
    file_handler.setLevel(logging.DEBUG)

    # create a console handler
    console_handler = logging.StreamHandler()

    # set it's level
    console_handler.setLevel(logging.WARNING)

    # create a formatter
    formatter = logging.Formatter(
        "%(asctime)s - %(levelname)s - %(message)s"
    )

    # give formatter to both handlers
    file_handler.setFormatter(formatter)
    console_handler.setFormatter(formatter)

    # attach handlers to the logger
    logger.addHandler(file_handler)
    logger.addHandler(console_handler)

    # log messages
    logger.debug("Debug message")

    logger.info("Info message")

    logger.warning("Warning message")

    logger.error("Error message")

    logger.critical("Critical message")


logging_in_file_and_console_updated()
```

##### If a logger's level is not set (NOTSET), it inherits its effective logging level from its parent logger (usually the root logger). Therefore, it does not automatically allow all messages. Setting logger.setLevel(logging.DEBUG) explicitly ensures the logger accepts all log levels (DEBUG and above), regardless of the root logger's configuration.

```python
# Assignment 3: Formatting Log Messages

# 1. Write a Python function to create a logger with a custom log message format that includes the timestamp, logging level, and message.

# logger does not format message, handler formats messages
def custom_logger():
    logger = logging.getLogger("custom_logger")
    logger.setLevel(logging.DEBUG)

    # prevent duplicates handlers in jupyter
    if logger.hasHandlers():
        logger.handlers.clear()

    # create a console handler
    console_handler = logging.StreamHandler()
    # The logger filters first, then the handler filters.

    # The effective level is therefore the stricter of the two.
    
    formatter = logging.Formatter(
        "%(asctime)s - %(levelname)s - %(message)s"
    )

    # attach formatter to handler
    console_handler.setFormatter(formatter)

    # attach handler to logger
    logger.addHandler(console_handler)

    logger.debug("Debug message")
    logger.info("Info message")
    logger.warning("Warning message")
    logger.error("Error message")
    logger.critical("Critical message")

custom_logger()
```

```python
# 2. Modify the function to use different formats for the file and console handlers.
def custom_logger_handlers():
    logger = logging.getLogger("custom_logger_handlers")
    logger.setLevel(logging.DEBUG)

    # prevent duplicates handlers in jupyter
    if logger.hasHandlers():
        logger.handlers.clear()

    # create handlers
    console_handler = logging.StreamHandler()
    file_handler = logging.FileHandler('app3.log')
    # The logger filters first, then the handler filters.

    # The effective level is therefore the stricter of the two.
    
    console_formatter = logging.Formatter(
        "%(asctime)s - %(levelname)s - %(message)s"
    )

    file_formatter = logging.Formatter(
        "%(levelname)s - %(message)s"
    )

    # set levels for handlers
    console_handler.setLevel(logging.DEBUG)
    file_handler.setLevel(logging.DEBUG)

    # attach formatter to handlers
    console_handler.setFormatter(console_formatter)
    file_handler.setFormatter(file_formatter)

    # attach handler to logger
    logger.addHandler(console_handler)
    logger.addHandler(file_handler)

    logger.debug("Debug message")
    logger.info("Info message")
    logger.warning("Warning message")
    logger.error("Error message")
    logger.critical("Critical message")

custom_logger_handlers()
```

```python
# Assignment 4: Rotating Log Files

# 1. Write a Python function to create a logger that uses a rotating file handler, which creates a new log file when the current log file reaches a certain size.
from logging.handlers import RotatingFileHandler
def rotating_logging():
    logger = logging.getLogger("rotating_logger")
    logger.setLevel(logging.DEBUG)

    # prevent duplicate handlers:
    if logger.hasHandlers():
        logger.handlers.clear()

    rotating_file_handler = RotatingFileHandler('rotating_app.log', maxBytes=2000, backupCount=5)
    rotating_file_handler.setLevel(logging.DEBUG)

    formatter = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(message)s')
    rotating_file_handler.setFormatter(formatter)
    logger.addHandler(rotating_file_handler)

    for i in range(120):
        logger.debug("This is a debug message number: {}".format(i))

rotating_logging()
```

```python
# Assignment 6: Contextual Logging

# 1. Write a Python function to create a logger that includes contextual information (e.g., function name, line number) in the log messages.
import logging
def contextual_logging():
    logger = logging.getLogger("contextual_logger")
    logger.setLevel(logging.DEBUG)

    if logger.hasHandlers():
        logger.handlers.clear()

    file_handler = logging.FileHandler('context_app.log')
    file_handler.setLevel(logging.DEBUG)

    formatter = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(message)s - %(funcName)s - %(lineno)d')
    file_handler.setFormatter(formatter)

    logger.addHandler(file_handler)

    def test_func():
            logger.debug('This is a debug message')
            logger.info('This is an info message')
            logger.warning('This is a warning message')
            logger.error('This is an error message')
            logger.critical('This is a critical message')
        
    test_func()


contextual_logging()
```

```python
# 2. Modify the function to include additional contextual information (e.g., user ID, session ID).
def updated_contextual_logging(user_id, session_id):
    logger = logging.getLogger("updated_contextual_logger")
    logger.setLevel(logging.DEBUG)
    
    if logger.hasHandlers():
        logger.handlers.clear()
    
    file_handler = logging.FileHandler('context_app1.log')
    file_handler.setLevel(logging.DEBUG)
    
    formatter = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(message)s - %(funcName)s - %(lineno)d - UserID: %(user_id)s - SessionID: %(session_id)s')
    file_handler.setFormatter(formatter)
    
    logger.addHandler(file_handler)

    # extra allows to add more data in log like user_id which is not already there in  python log recrod
    extra = {'user_id': user_id, 'session_id': session_id}
    
    def test_func():
        logger.debug('This is a debug message', extra=extra)
        logger.info('This is an info message', extra=extra)
        logger.warning('This is a warning message', extra=extra)
        logger.error('This is an error message', extra=extra)
        logger.critical('This is a critical message', extra=extra)
            
    test_func()
    
    
updated_contextual_logging(123, 456)
```

```python
# Assignment 7: Configuring Logging with a Dictionary

# 1. Write a Python function to configure logging using a dictionary. The configuration should include handlers for both file and console logging.
import logging
import logging.config
def logging_with_dictionary():
    log_config = {
        'version': 1, # currently python supports only version 1
        'formatters': {
            # formatter 1
            # default is just a name and could be anything
            'default': {
                'format': '%(asctime)s - %(name)s - %(levelname)s - %(message)s'
            },
            # formatter 2
            'detailed': {
                'format': '%(asctime)s - %(name)s - %(levelname)s - %(message)s - %(funcName)s - %(lineno)d'
            }
        },
        'handlers': {
            'file': {
                'class': 'logging.FileHandler',
                'filename': 'dict_config_app.log',
                'formatter': 'detailed',
                'level': 'DEBUG',
            },
            'console': {
                'class': 'logging.StreamHandler',
                'formatter': 'default',
            }
        },
        'root': {
            'handlers': ['file', 'console'],
            'level': 'DEBUG'
        }
    }

    logging.config.dictConfig(log_config)
    logger = logging.getLogger('')
    logger.debug('This is a debug message')
    logger.info('This is an info message')
    logger.warning('This is a warning message')
    logger.error('This is an error message')
    logger.critical('This is a critical message')


logging_with_dictionary()
```

```python
# Assignment 9: Logging Performance

# 1. Write a Python script to benchmark the performance of logging with different handlers (e.g., file handler, console handler, rotating file handler).
import time
from logging.handlers import RotatingFileHandler
def benchmark_with_handlers():
    logger = logging.getLogger("benchmark_logger")
    logger.setLevel(logging.DEBUG)

    if logger.hasHandlers():
        logger.handlers.clear()

    # file handler
    file_handler = logging.FileHandler('benchmark_app.log')
    file_handler.setLevel(logging.DEBUG)
    logger.addHandler(file_handler)

    start_time = time.time()
    for i in range(10000):
        logger.debug('This is a debug message')
    end_time = time.time()
    print('File handler logging time: {}s'.format(end_time - start_time))
    logger.removeHandler(file_handler)

    # console handler
    console_handler = logging.StreamHandler()
    console_handler.setLevel(logging.DEBUG)
    logger.addHandler(console_handler)

    start_time = time.time()
    for i in range(10000):
        logger.debug('This is a debug message')
    end_time = time.time()
    print('Console handler logging time: {}s'.format(end_time - start_time))
    logger.removeHandler(console_handler)

    # rotating file handler
    rotating_handler  = RotatingFileHandler('rotating_benchmark.log', maxBytes=2000, backupCount=5)
    rotating_handler.setLevel(logging.DEBUG)
    logger.addHandler(rotating_handler)

    start_time = time.time()
    for i in range(10000):
        logger.debug('This is a debug message')
    end_time = time.time()
    print('Rotating handler logging time: {}s'.format(end_time - start_time))
    logger.removeHandler(rotating_handler)

benchmark_with_handlers()
    
```

```python
# 2. Modify the script to compare the performance of logging with and without message formatting.
import time
from logging.handlers import RotatingFileHandler
def benchmark_with_handlers_with_formatting():
    logger = logging.getLogger("benchmark_logger_formatting")
    logger.setLevel(logging.DEBUG)

    if logger.hasHandlers():
        logger.handlers.clear()

    # file handler without formatting
    file_handler = logging.FileHandler('benchmark_app_no_format.log')
    file_handler.setLevel(logging.DEBUG)
    logger.addHandler(file_handler)

    start_time = time.time()
    for i in range(1000):
        logger.debug('This is a debug message')
    end_time = time.time()
    print('File handler logging time without formatting: {}s'.format(end_time - start_time))
    logger.removeHandler(file_handler)

    # file handler with formatting
    file_handler = logging.FileHandler('benchmark_app_with_format.log')
    file_handler.setLevel(logging.DEBUG)
    formatter = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(message)s')
    file_handler.setFormatter(formatter)
    logger.addHandler(file_handler)

    start_time = time.time()
    for i in range(1000):
        logger.debug('This is a debug message')
    end_time = time.time()
    print('File handler logging time with formatting: {}s'.format(end_time - start_time))
    logger.removeHandler(file_handler)

benchmark_with_handlers_with_formatting()
    
```
